```python

scores = [99, 98, 97, 96, 80]

#Display score
def display_scores():
  print("The scores are")
for i in range(len(scores)):
  print(scores[i])

#search for score
def search_score():
  search = int(input("What number to search for?", ))
  count =scores.count(search)
  print("number of", search,"counted is", count)

#compute the average of the score
def compute_average_score():
  average = (sum(scores)/len(scores))
  print(average)

#update score
def update_score():
  index = int(input("What index do we update??"))
  new_score = int(input("Enterr the new score"))
  scores[index] = new_score
  print("score updated")

def print_update():
  for i in range (len(scores)):
    print("Student", i +  1, ":", scores[1])


while True:
  print("Althea's Student Score Analyzer")
  print("1. Display scores")
  print("2. Search scores")
  print("3. Find average scores")
  print("4. Update scores")
  print("5. Print updated score")
  print("6. Exit")

  choice = input("Choose a programmie(1-6): ")

  if choice == "1":
    display_scores()

  elif choice == "2":
    search_score()

  elif choice == "3":
    compute_average_score()

  elif choice == "4":
    update_score()

  elif choice == "5":
    print_update()

  else:
    print("Bye best friend")
    break



```
